#### Логическая индексация.

Для поиска в данных можно использовать более сложные конструкции. Пример ниже выводит все записи о золотых медалях. Можно использовать любые функции сравнения, логические функции и пр.

```python
data[data['Medal'] == 'Gold']
```

Функции статистики позволяют вывести минимальные, максимальные, средние значения, сумму значений и некоторые другие характеристики (см. документацию).

Пример ниже выводит масимальный и минимальный возраст золотых медалистов за все время проведения олимпийских игр:

```python
data[data['Medal'] == 'Gold']['Age'].max(), \
data[data['Medal'] == 'Gold']['Age'].min()
```

Если необходимо вывести несколько максимальных (или минимальных) значений - можно использовать фунцию nlargest (nsmallest), которая выводит n самых больших значений в столбце.

```python
data[data['Medal'] == 'Gold'].nlargest(3, 'Age')
```

Как видно, функция не отбрасывает повторяющиеся значения. Если мы хотим узнать возраст самых старших спорстменов с привязкой не к имени, а к возрасту, нужно использовать функцию drop_dublicates:

```python
data[data['Medal'] == 'Gold'].drop_duplicates('Age').nlargest(3,'Age')
```

Если передать функции nlargest один столбец, второй параметр функции станет необязательным.

Средний возраст медалистов по видам медалей:

```python
data[data['Medal'] == 'Gold']['Age'].mean(), \
data[data['Medal'] == 'Silver']['Age'].mean(), \
data[data['Medal'] == 'Bronze']['Age'].mean()
```

Для формирования более сложного запроса используем логический оператор "И". Определим средний возраст медадистов из России:

```python
data[(data['Medal'] == 'Gold') & (data['Team'] == 'Russia')]['Age'].mean(), \
data[(data['Medal'] == 'Silver') & (data['Team'] == 'Russia')]['Age'].mean(), \
data[(data['Medal'] == 'Bronze') & (data['Team'] == 'Russia')]['Age'].mean()
```

Попробуем передать в качестве запроса результат другого запроса. Попробуйте разобраться с конструкцией ниже, после прочитайте объяснение.

```python
data[(data['Age'] == data[data['Medal'] == 'Gold']['Age'].max()) & (data['Medal'] == 'Gold')]
```

В показанном примере запрос data[data['Medal'] == 'Gold']['Age'].max() максимальный возраст золотого медалиста - 64. Это число указывается в запросе по столбцу Age. То есть: data[(data['Age'] == 64). После этого при помощи оператора И отбираются только золотые медалисты, имеющие указанный возраст. 

Добавив в конце оператор метод nunique() мы получим статистику количества уникальных записей по столбцам таблицы:

```python
data[(data['Age'] == data[data['Medal'] == 'Gold']['Age'].max()) & (data['Medal'] == 'Gold')].nunique()
```

Из этой таблицы мы видим что в полученном DataFrame есть 2 записи (ID у всех уникальный), это 2 разных спортсмена (2 имени), одного позраста и пола и. т. д.

Для того, чтобы в качетве ответа получить число, ограничим запрос на вывод только столбца Name, в котором и посчитаем уникальные записи:

```python
data[(data['Age'] == data[data['Medal'] == 'Gold']['Age'].max()) & (data['Medal'] == 'Gold')]['Name'].nunique()
```

### Запросы (query)

Как вы могли заметить, порой запросы получаются довольно длинными и сложными, и, как вы убедитесь во время самостоятельной практики, ошибиться здесь довольно просто. Для облегчения визуального восприятия, а так же для програмного формирования текстовых запросов предусмотрен метод query, который принимает предварительно сформированный текстовый запрос.

Вывод всех мужчин-футболистов олимпиады 2008:

```python
q = '(Year == 2008) & (Sex == "M") & (Sport == "Football")'
data.query(q).head()
```

Вывод количества золотых медалистов сборной России, не достигших 18 лет:

```python
q = '(Team == "Russia") & (Age < 18) & (Medal == "Gold")'
data.query(q).shape[0]
```

В предыдущем примере есть ошибка. Несовершеннолетних автлетов не 16, а 15. Найдите ошибку и исправьте запрос.

Поработаем еще с этим запросом. Вывод первых трех строк таблицы:

```python
data.query(q).head(3)
```

Определим средний возраст этих молодых призеров:

```python
data.query(q).Age.mean()
```

### Группировка (GroupBy)

Группировка позволяет сгруппировать данные по столбцам, имеющим булевский или числовой тип и вывести число записей по каждой группе:

```python
data.groupby('Sex')['Age', 'Height', 'Weight'].count()
```

Для числовых значений можно использовать статистические функции с помощью функции аггрегатора agg:

```python
data.groupby('Sex')['Age', 'Height', 'Weight'].agg(np.mean)
```

```python
data.groupby('Sex')['Age', 'Height', 'Weight'].agg(np.max)
```

### Сводная таблица (CrossTab) 
это продвинутый инструмент формирования сводных таблиц. Ознакомьтесь с документацией для ознакомления с возможностями этого инструмента. Пример: выведем статистику завоеванных медалей для атлетов младше 18 лет:

```python
pd.crosstab(data[data['Age'] < 18]['Age'], data['Medal'])
```

Статистика медалей по странам за все время:

```python
pd.crosstab(data['Team'], data['Medal']).head(3)
```
